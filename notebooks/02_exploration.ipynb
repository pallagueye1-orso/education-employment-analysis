# 02 — Exploration des données nettoyées

**Objectif** : Vérifier la qualité et comprendre la structure des données nettoyées avant de passer aux analyses.  
Ce notebook ne produit pas de fichier — il sert uniquement à la vérification et à la découverte.

**Fichiers en entrée** :
- `data/clean/emploi_clean.csv`
- `data/clean/chomage_clean.csv`

```python
import pandas as pd
import numpy as np
from pathlib import Path

# Chemin du projet et des données
PROJECT_ROOT = Path(r'C:\Users\HP\Documents\education-employment-analysis')
CLEAN_DIR = PROJECT_ROOT / 'data' / 'clean'

# Chargement des fichiers nettoyés
emploi  = pd.read_csv(CLEAN_DIR / 'emploi_clean.csv')
chomage = pd.read_csv(CLEAN_DIR / 'chomage_clean.csv')

print('Fichiers chargés.')
print(f'  emploi  : {emploi.shape}')
print(f'  chomage : {chomage.shape}')
```

## 1. Aperçu général

```python
print('=== EMPLOI ===')
display(emploi.head(6))
print()
print('Types de colonnes :')
print(emploi.dtypes)
```

```python
print('=== CHÔMAGE ===')
display(chomage.head(6))
print()
print('Types de colonnes :')
print(chomage.dtypes)
```

## 2. Couverture géographique et temporelle

```python
print('Couverture emploi :')
print(f'  Pays  : {emploi["country"].nunique()}')
print(f'  Années: {emploi["year"].min()} – {emploi["year"].max()}')
print(f'  Observations par niveau :')
print(emploi.groupby('education_level')['country'].count().reindex(['LOW','MEDIUM','HIGH']))
print()
print('Couverture chômage :')
print(f'  Pays  : {chomage["country"].nunique()}')
print(f'  Années: {chomage["year"].min()} – {chomage["year"].max()}')
print(f'  Observations par niveau :')
print(chomage.groupby('education_level')['country'].count().reindex(['LOW','MEDIUM','HIGH']))
```

```python
# Quelques pays disponibles (30 premiers par ordre alphabétique)
print('Exemples de pays dans le dataset emploi :')
print(sorted(emploi['country'].unique())[:30])
```

## 3. Statistiques descriptives

```python
print('Statistiques nb_employes (en milliers) par niveau :')
display(
    emploi.groupby('education_level')['nb_employes']
    .describe()
    .round(1)
    .reindex(['LOW','MEDIUM','HIGH'])
)
```

```python
print('Statistiques nb_chomeurs (en milliers) par niveau :')
display(
    chomage.groupby('education_level')['nb_chomeurs']
    .describe()
    .round(1)
    .reindex(['LOW','MEDIUM','HIGH'])
)
```

## 4. Vérification de la compatibilité pour la fusion

```python
# Test du inner join : combien de paires (country, year, education_level) sont communes ?
fusion_test = pd.merge(
    emploi[['country', 'year', 'education_level']],
    chomage[['country', 'year', 'education_level']],
    on=['country', 'year', 'education_level'],
    how='inner'
)

print('=== COMPATIBILITÉ POUR LA FUSION ===')
print(f'Lignes dans emploi           : {len(emploi)}')
print(f'Lignes dans chômage          : {len(chomage)}')
print(f'Lignes après inner join      : {len(fusion_test)}')
print(f'Pays couverts après fusion   : {fusion_test["country"].nunique()}')
print()
print('Lignes par niveau après fusion :')
print(fusion_test['education_level'].value_counts().reindex(['LOW','MEDIUM','HIGH']))
```

## 5. Points d'attention notés

Ce notebook permet d'identifier les particularités à mentionner dans le rapport :
- La valeur `nb_employes` et `nb_chomeurs` est exprimée en **milliers de personnes**.
- Le dataset ne contient que **3 niveaux d'éducation agrégés** (LOW / MEDIUM / HIGH) pour maximiser la couverture géographique (codes AGGREGATE ILO).
- L'analyse sur les diplômes universitaires détaillés (ISCED 5 à 8) sera faite dans `03_analyses.ipynb` avec un sous-ensemble de pays.
- Le taux de chômage est **calculé** (non fourni directement) : `taux = nb_chomeurs / (nb_employes + nb_chomeurs) × 100`.
