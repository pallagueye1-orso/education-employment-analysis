```
import pandas as pd

df = pd.read_csv("EMP_TEMP_SEX_EDU_NB_A-20260409T2352.csv")
df.head()
```

```
# On verifie les années.
df["time"].unique()
```

```
# On contrôle tous les colonnes qu'on a dans la base.
df.columns
```

```
# On garde que les colonnes utiles.
df = df[[
    "ref_area",     # pays
    "time",         # année
    "sex",
    "classif1",     # éducation
    "obs_value",    # valeur
    "indicator"     # type (emploi)
]]
```

```
## Pas de : df = df[df["time"] >= 2000], sinon on perdrait
# trop d'infos en filtrant l'année de départ.

## PAS DE : df = df.dropna(),
# sinon on perdrait trop de données.
```

```
# On renomme pour que ça soit plus lisible
df = df.rename(columns={
    "ref_area": "country",
    "time": "year",
    "classif1": "education",
    "obs_value": "value"
})
```

```
# On traduit aussi l'éducation
def simplify_edu(x):
    x = str(x)
    if "0" in x or "1" in x or "2" in x:
        return "LOW"
    elif "3" in x or "4" in x:
        return "MEDIUM"
    else:
        return "HIGH"

df["education_level"] = df["education"].apply(simplify_edu)
```

```
# On verifie
df.shape
df["year"].min(), df["year"].max()
```

```
df = df[df["year"] >= 2000]
```

```
df["year"].min(), df["year"].max()
```

```
# On sauvegarde la nouvelle base
df.to_csv("ilo_cleaned_final.csv", index=False)
```
